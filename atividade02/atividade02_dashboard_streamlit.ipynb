## Objetivo

Dar continuidade à Atividade 01 transformando a proposta de análise visual em um dashboard local para explorar preços, volume e setores das empresas do S&P 500.

## Perguntas da Atividade 01 contempladas

1. Como o preço médio de fechamento varia ao longo dos anos para diferentes empresas?
2. Quais empresas apresentam maior volume médio de negociação?
3. Como as empresas estão distribuídas por setor?

## Visualizações e bibliotecas

O dashboard utiliza três visualizações: uma linha temporal em Altair, um gráfico de barras de volume em Altair e um gráfico de barras de empresas por setor em Plotly. Os gráficos Altair possuem tooltips com os valores das marcas. Streamlit é usado para estruturar a aplicação e os controles.

## Controles interativos

O usuário pode selecionar o setor e o intervalo de datas. Para cada setor, o dashboard exibe automaticamente os cinco `symbol` com maior volume médio de negociação no conjunto completo. As visualizações são atualizadas conforme os filtros. Quando o recorte não possui registros, o dashboard informa a situação.

## Performance

A tabela completa de cotações é carregada no pandas, sem amostragem. `usecols`, `dtype`, `st.cache_resource` e um índice ordenado por `symbol` reduzem o custo das interações: a carga e o cálculo das cinco empresas por setor ocorrem apenas uma vez.

## Resultado observado e limitação

A aplicação permite comparar a evolução anual dos preços e os volumes médios das empresas selecionadas. Nesta etapa, não foi realizada limpeza, imputação ou modelagem preditiva; as análises são exploratórias.

## Execução

Consulte o `README.md` e execute `streamlit run app.py`.
